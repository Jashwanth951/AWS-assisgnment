```python
plt.figsize=(10, 6)
sns.countplot(y='profession', data=dataset, order=dataset['profession'].value_counts().index)
plt.title('Distribution of Professions')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(dataset['ease_of_use'], bins=10, kde=True)
plt.title('Distribution of ease of use')
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='ease_of_use', y='profession', data=dataset)
plt.title('ease of use by Profession')
plt.show()
```
